```python
 #1 Social Media – Post Engagement Analyzer

 def post_engagement(likes_list):
    total_likes = sum(likes_list)

    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    print("Total Likes:", total_likes)
    print("Post Status:", status)

# Example
post_engagement([200, 300, 250, 300])
```

```python
#2  Healthcare – Medicine Stock Alert System

def medicine_stock(stock):
    if stock < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"

    print("Medicine Stock:", stock)
    print("Status:", status)

# Example
medicine_stock(6)
```

```python
#3 Agriculture – Rainfall Adequacy Checker

def rainfall_checker(rainfall_list, required_level):
    average = sum(rainfall_list) / len(rainfall_list)

    if average >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print("Average Rainfall:", average)
    print("Rainfall Status:", status)

# Example
rainfall_checker([70, 75, 71], 70)
```

```python
#4 Social Media – Duplicate Account Detection

def check_duplicates(usernames):
    if len(usernames) != len(set(usernames)):
        result = "Yes"
    else:
        result = "No"

    print("Duplicate Accounts Found:", result)

# Example
check_duplicates(["harini", "raj", "harini"])
```

```python
#5 Healthcare – Appointment Eligibility Checker

def appointment_eligibility(age):
    if age >= 18:
        status = "Eligible"
    else:
        status = "Not Eligible"

    print("Patient Age:", age)
    print("Eligibility Status:", status)

# Example
appointment_eligibility(21)
```

```python
#6 Agriculture – Premium Crop Price Filter

def premium_crops(prices):
    premium = [price for price in prices if price > 2000]

    print("Premium Crops:", premium)

# Example
premium_crops([1500, 2500, 3200, 1800])
```

```python
#7 System Monitoring – Application Health Checker

def app_health(errors):
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print("Error Count:", errors)
    print("System Status:", status)

# Example
app_health(7)
```

```python
#8 Banking – Daily Transaction Limit Checker

def transaction_checker(amount):
    limit = 50000

    if amount <= limit:
        status = "Approved"
    else:
        status = "Rejected"

    print("Transaction Amount:", amount)
    print("Transaction Status:", status)

# Example
transaction_checker(60000)
```

```python
#9 E-Learning – Student Attendance Eligibility System

def attendance_checker(attended, total):
    percentage = (attended / total) * 100

    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"

    print("Attendance Percentage:", percentage)
    print("Exam Eligibility:", status)

# Example
attendance_checker(40, 50)

```

```python
#10 Smart Electricity Bill Analyzer

def electricity_bill(units):
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    print("Units Consumed:", units)
    print("Total Bill:", bill)
    print("Usage Classification:", usage)

# Example
electricity_bill(250)
```
