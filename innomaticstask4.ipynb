```python
#Problem 1: Mobile Recharge Validation System


def validate_recharge(amount):
    valid_plans = [199, 299, 399, 599]

    if amount < 50:
        print("Recharge amount must be at least ₹50")
        return False
    elif amount not in valid_plans:
        print("Invalid plan selected")
        return False
    else:
        print("Recharge Successful ✅")
        return True


# Retry option using while loop
while True:
    amt = int(input("Enter recharge amount: "))
    if validate_recharge(amt):
        break

```

```python
#Problem 2: Inventory Reorder Alert System


def check_inventory(products):
    for item, stock in products.items():
        if stock < 15:
            print(item, ": Reorder Alert ⚠")
        else:
            print(item, ": Stock OK")


inventory = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 12,
    "Monitor": 20
}

check_inventory(inventory)
```

```python
#Problem 3: Student Result Processing System

def process_result(marks):
    total = 0

    for m in marks:
        total += m

    average = total / len(marks)

    print("Average:", average)

    if average >= 50:
        return "Pass"
    else:
        return "Fail"


marks_list = [60, 45, 70, 50]
result = process_result(marks_list)
print("Result:", result)
```

```python
#Problem 4: Cab Fare Estimator with Retry Option

def calculate_fare(distance, peak):
    fare = 50 + (12 * distance)

    if peak == "yes":
        fare = fare + (fare * 0.25)

    return fare


while True:
    km = float(input("Enter distance (km): "))
    peak_hour = input("Is it peak hour? (yes/no): ")

    total_fare = calculate_fare(km, peak_hour)
    print("Total Fare: ₹", total_fare)

    retry = input("Do you want to calculate again? (yes/no): ")
    if retry == "no":
        break
```

```python
#Problem 5: Employee Attendance Eligibility Checker

def check_attendance(att_list):
    present_days = 0

    for day in att_list:
        if day == "P":
            present_days += 1

    percentage = (present_days / len(att_list)) * 100

    print("Attendance %:", percentage)

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


attendance = ["P", "A", "P", "P", "A", "P", "P"]
status = check_attendance(attendance)
print("Status:", status)
```

```python
#Problem 6: Password Strength Checker

def check_password(password):
    has_digit = False
    has_special = False

    if len(password) < 8:
        return "Weak Password ❌ (Too Short)"

    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in "@#$":
            has_special = True

    if has_digit and has_special:
        return "Strong Password ✅"
    else:
        return "Weak Password ❌"


pwd = input("Enter password: ")
print(check_password(pwd))
```
