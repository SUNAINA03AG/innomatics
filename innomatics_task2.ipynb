```python
#Problem 1: Unique Words in a Sentence

sentence=input("Enter Sentence")

words=sentence.split()
unique_words=set(words)
count=len(unique_words)
print("Unique words count:",count )
print("Unique words",unique_words)
```

```python
#Problem 2: Highest Salary from Employee Data

employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

highest_emp=max(employees,key=employees.get)

print(f"Highest Salary:{highest_emp}-{employees[highest_emp]}")
```

```python
#Problem 3: Find Maximum and Minimum Values

numbers = [45, 22, 89, 10, 66]

min_val=numbers[0]
max_val=numbers[0]

for i in numbers:
    if i > max_val:
        max_val=i

    if i < min_val:
        min_val=i

print("List",numbers)
print("Max Value",max_val)
print("Min Value",min_val)
```

```python
#Problem 4: Count Products Above 1000

prices = [450, 1200, 899, 1500, 300]
count=0
for i in prices:
    if i > 1000:
        count += 1
    else:
        pass

print(f"Products above 1000:{count}")
```

```python
#Problem 5: Calculate Attendance Percentage

attendance = ["P", "P", "A", "P", "P"]

count=attendance.count("P")
total=len(attendance)

per=(count/total)*100

print("Attendance Percentage:", per)
```

```python
#Problem 6: Remove Duplicate Phone Numbers


phone_numbers = [9876543210, 9123456789, 9876543210]
duplicate=set(phone_numbers)

print(f"Unique phone numbers:{duplicate}")
```

```python
# Problem 7: Count Character Frequency

text = "python"

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
```

```python
# Problem 9: Check if Key Exists in Dictionary

employees = {
    "ravi": 75000,
    "anita": 68000
}
# key="ravi"
key=input("Enter a key:")

if key in employees:
    print("Employee Exists")

else:
    print("Employee does not exist")
```

```python
# Problem 10: Calculate Average Marks

marks=[70,75,80,65]

total=sum(marks)
avg=total/len(marks)
print("Average Marks:", avg)


count=0
for i in marks:
    count += i

avg=count/len(marks)
print("Average Marks:", avg)
```
