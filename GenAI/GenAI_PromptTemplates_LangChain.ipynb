# **Replace Hardcoded Prompt**

```python
prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

output = prompt.format(topic="Artificial Intelligence")
print(" Output:", output)
```

# **Multi-Input Prompt System**

```python
multi_prompt = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

print("Output:", multi_prompt.format(
    topic="Python",
    audience="kids",
    tone="fun"
))
```

# **Prompt Variations Engine**

```python
teaching_prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

interview_prompt = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 questions about {topic}"
)

storytelling_prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

print(" Teaching:", teaching_prompt.format(topic="Machine Learning"))
print(" Interview:", interview_prompt.format(topic="Machine Learning"))
print("Storytelling:", storytelling_prompt.format(topic="Machine Learning"))
```

# **ChatPromptTemplate System**

```python
chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", "Explain {topic} in simple terms.")
])

messages = chat_template.format_messages(
    role="teacher",
    topic="Neural Networks"
)

print(" Output:")
for msg in messages:
    print(msg.type, ":", msg.content)
```

# **Input Validation Layer**

```python
def validate_inputs(audience, tone):
    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        return "Invalid audience!"
    if tone not in valid_tone:
        return "Invalid tone!"
    return "Valid input"

print(" Valid Case:", validate_inputs("beginner", "fun"))
print("Invalid Case:", validate_inputs("expert", "boring"))
```

# **Prompt Generator App**

```python
generator_template = PromptTemplate(
    input_variables=["topic", "audience", "tone", "style"],
    template="Explain {topic} for {audience} in a {tone} tone using {style} style"
)

def generate_prompt(topic, audience, tone, style):
    # Validate input
    validation = validate_inputs(audience, tone)
    if validation != "Valid input":
        return validation

    valid_styles = ["teaching", "interview", "storytelling"]
    if style not in valid_styles:
        return "Invalid style!"

    return generator_template.format(
        topic=topic,
        audience=audience,
        tone=tone,
        style=style
    )

print(" Output:", generate_prompt("Neural Networks", "beginner", "fun", "storytelling"))
```

# **Template Reusability Test**

```python
reusable_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

test_inputs = [
    {"topic": "AI", "audience": "beginner", "tone": "fun"},
    {"topic": "Deep Learning", "audience": "intermediate", "tone": "formal"},
    {"topic": "NLP", "audience": "expert", "tone": "casual"},
    {"topic": "Computer Vision", "audience": "beginner", "tone": "formal"},
    {"topic": "Data Science", "audience": "intermediate", "tone": "fun"}
]

print(" Outputs:")
for inp in test_inputs:
    print(reusable_template.format(**inp))
```

# **Final Pipeline Flow**

```python
# Final execution pipeline

topic = "Neural Networks"
audience = "beginner"
tone = "fun"
style = "storytelling"

# Generate final prompt
final_result = generate_prompt(topic, audience, tone, style)

print("Final Output:")
print(final_result)
```
