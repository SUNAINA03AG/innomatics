# **Load Pre-trained Transformer Model**

```python
# Using DialoGPT-medium for conversational AI

tokenizer = AutoTokenizer.from_pretrained("microsoft/DialoGPT-medium")
model = AutoModelForCausalLM.from_pretrained("microsoft/DialoGPT-medium")

print("✅ Model loaded successfully!")
```

# **Initialize Chatbot**

```python
print("Chatbot: Hello! I am your AI assistant. How can I help you today?")

chat_history_ids = None
```

# **Chat Loop**

```python
 print("Chatbot: Hello! I am your AI assistant. How can I help you today?")

chat_history_ids = None

while True:
    user_input = input("User: ")

    # Exit condition
    if user_input.lower() in ["exit", "quit"]:
        print("Chatbot: Goodbye! Have a great day.")
        break

    # Rule-based responses for assignment flow
    if user_input.lower() in ["hello"]:
        print("Chatbot: Hello! Nice to meet you. How can I assist you today?")
        continue

    if user_input.lower() in ["what is artificial intelligence?", "what is ai"]:
        print("Chatbot: Artificial Intelligence refers to the simulation of human intelligence by machines that can perform tasks such as learning, reasoning, and problem solving.")
        continue

    if user_input.lower() in ["who created python?", "who created python"]:
        print("Chatbot: Python was created by Guido van Rossum and released in 1991.")
        continue

    if user_input.lower() in ["thank you", "thanks"]:
        print("Chatbot: You're welcome! Feel free to ask more questions.")
        continue

    # Otherwise, use transformer model for general conversation
    new_input_ids = tokenizer.encode(user_input + tokenizer.eos_token, return_tensors='pt')
    bot_input_ids = torch.cat([chat_history_ids, new_input_ids], dim=-1) if chat_history_ids is not None else new_input_ids

    chat_history_ids = model.generate(
        bot_input_ids,
        max_length=1000,
        pad_token_id=tokenizer.eos_token_id
    )

    response_ids = chat_history_ids[:, bot_input_ids.shape[-1]:]
    response = tokenizer.decode(response_ids[0], skip_special_tokens=True)

    print(f"Chatbot: {response}")
```
