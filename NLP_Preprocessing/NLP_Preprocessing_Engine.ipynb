# **Task 2 :Build Advanced Preprocessing Function**

```python
import re
from collections import Counter

def preprocess_text(text):
    if not text or text.strip() == "":
        return [], ""

    text = text.lower()

    # Remove URLs and emails
    text = re.sub(r'http\S+|www\S+|\S+@\S+', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Handle repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1', text)

    # Remove special characters (including emojis)
    text = re.sub(r'[^a-z\s]', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenization
    tokens = text.split()

    # Remove short words (≤2 except no, not)
    tokens = [t for t in tokens if len(t) > 2 or t in ["no", "not"]]

    cleaned_sentence = " ".join(tokens)

    return tokens, cleaned_sentence
```

# **Task 3: Stress Testing**

```python
sample_inputs = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

all_tokens = []

for s in sample_inputs:
    tokens, cleaned = preprocess_text(s)
    all_tokens.extend(tokens)

    print("Original:", s)
    print("Tokens:", tokens)
    print("Cleaned Sentence:", cleaned)
    print()
```

# **Task 4: Token Analytics**

```python
for s in sample_inputs:
    tokens, _ = preprocess_text(s)

    total = len(tokens)
    unique_tokens = len(set(tokens))
    avg_len = sum(len(w) for w in tokens) / total if total > 0 else 0

    print("Sentence:", s)
    print("Total Tokens:", total)
    print("Unique Tokens:", unique_tokens)
    print("Avg Token Length:", round(avg_len, 2))
    print()

# Analysis Answers
print("Analysis:")
print("Most noisy sentence: 'Get 100% FREE access now!!!'")
print("Most meaningful after cleaning: 'I am not happy with this'")
```

# **Task 5: Frequency Analysis**

```python
counter = Counter(all_tokens)

print("Top 10 Frequent Words:")
print(counter.most_common(10))

print("\nTop 5 Least Frequent Words:")
print(counter.most_common()[-5:])
```

# **Task 6: Build Full Pipeline**

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, cleaned = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(cleaned)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

result = full_pipeline(sample_inputs)

print(result)
```
