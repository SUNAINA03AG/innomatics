# **Load Pre-trained Transformer Model (DialoGPT)**

````python
# Step 2: Load Pre-trained Transformer Model (DialoGPT)
tokenizer = AutoTokenizer.from_pretrained("microsoft/DialoGPT-medium")
model = AutoModelForCausalLM.from_pretrained("microsoft/DialoGPT-medium")

print("Chatbot: Hello! I am your AI assistant. How can I help you today?\n")
````

# **Conversation Loop**

````python
# Step 3: Conversation Loop
while True:
    # Accept User Input
    user_input = input("User: ")

    # Exit Condition
    if user_input.lower() in ["exit", "quit"]:
        print("Chatbot: Goodbye! Have a great day.")
        break

    # Encode User Input
    input_ids = tokenizer.encode(user_input + tokenizer.eos_token, return_tensors='pt')

    # Create attention mask
    attention_mask = torch.ones(input_ids.shape, dtype=torch.long)

    # Generate Response
    bot_output = model.generate(
        input_ids=input_ids,
        max_length=200,
        pad_token_id=tokenizer.eos_token_id,
        attention_mask=attention_mask
    )

    # Decode Response
    response = tokenizer.decode(bot_output[:, input_ids.shape[-1]:][0], skip_special_tokens=True)

    # Rule-based Overrides (Assignment Requirement)
    if "who created python" in user_input.lower():
        response = "Python was created by Guido van Rossum and released in 1991."
    elif "thank you" in user_input.lower():
        response = "You're welcome! Feel free to ask more questions."
    elif "what is ai" in user_input.lower() or "what is artificial intelligence" in user_input.lower():
        response = "Artificial Intelligence refers to the simulation of human intelligence by machines that can perform tasks such as learning, reasoning, and problem solving."

    print(f"Chatbot: {response}")
````
