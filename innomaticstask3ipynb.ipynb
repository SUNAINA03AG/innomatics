```python
#Problem 1: Employee Performance Bonus Eligibility

employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

highest_score = max(employees.values())

top_performers = []
for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)

print("Top Performers Eligible for Bonus:", ", ".join(top_performers), f"(Score: {highest_score})")
```

```python
#Problem 2: Search Query Keyword Analysis


i="Buy mobile phone buy phone online"
i=i.lower()
words=i.split()

fre={}

for n in words:
    if n in fre:
        fre[n] +=1
    else:
        fre[n] = 1

res={}
for n,c in fre.items():
    if c>1:
        res[n]=c
print(res)
```

```python
#Problem 3: Sensor Data Validation

sensor_readings = [3, 4, 7, 8, 10, 12, 5]
valid_readings = []

for i,val in enumerate(sensor_readings):
    if val % 2== 0:
         valid_readings.append((i, val))

print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

```python
# Problem 4: Email Domain Usage Analysis

emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

count = {}

for i in emails:
    domain = i.split("@")[1]
    if domain in count:
        count[domain] += 1
    else:
        count[domain] = 1

total = len(emails)

for domain, val in count.items():
    per = (val / total) * 100
    print(f"{domain}: {per:.0f}%")
```

```python
# Problem 5: Sales Spike Detection

sales = [1200, 1500, 900, 2200, 1400, 3000]

total = 0
for s in sales:
    total += s
l=len(sales)
avg = total / l

limit = avg * 1.3

for i in range(l):
    if sales[i] > limit:
        print(f"Day {i+1}: {sales[i]}")
```

```python
# Problem 6: Duplicate User ID Detection

user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

user_count = {}

for u in user_ids:
    if u in user_count:
        user_count[u] += 1
    else:
        user_count[u] = 1

for u, count in user_count.items():
    if count > 1:
        print(f"{u} → {count} times")
```
